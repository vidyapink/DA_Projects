#### Data type correction

```python
df["Release_Date"]=pd.to_datetime(df["Release_Date"],dayfirst=True,errors="coerce")
df["Vote_Count"]=pd.to_numeric(df["Vote_Count"],errors="coerce")
```

#### Group by
